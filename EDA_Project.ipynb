```python
Q1 = df['Billing Amount'].quantile(0.25)

Q3 = df['Billing Amount'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```


## 12. Outlier Analysis

Outlier analysis is performed to identify unusually high
or low observations in numerical variables.

In this project, the Interquartile Range (IQR) method
is used to identify potential outliers in Billing Amount
and Length of Stay.

Boxplots are used to visualize the distribution and
identify observations outside the IQR boundaries.

The identified outliers are investigated rather than
automatically removed.


```python
billing_outliers = df[
    (df['Billing Amount'] < lower_limit) |
    (df['Billing Amount'] > upper_limit)
]

print("Total Billing Outliers:", len(billing_outliers))
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(x=df['Length of Stay'])

plt.title('Outlier Detection in Length of Stay')
plt.xlabel('Length of Stay (Days)')

plt.show()
```

```python
Q1 = df['Length of Stay'].quantile(0.25)

Q3 = df['Length of Stay'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

stay_outliers = df[
    (df['Length of Stay'] < lower_limit) |
    (df['Length of Stay'] > upper_limit)
]

print("Total Length of Stay Outliers:", len(stay_outliers))
```

```python
print("Billing Outliers:", len(billing_outliers))

print("Length of Stay Outliers:", len(stay_outliers))
```


### Observation: Outlier Analysis

The IQR method was used to identify potential outliers
in Billing Amount and Length of Stay.

Boxplots were created to examine the distribution
of both numerical variables.

The analysis helps identify unusual observations
that may require further investigation.

Outliers were not automatically removed because
extreme values may represent genuine patient records.
